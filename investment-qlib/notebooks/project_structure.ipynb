## 目录说明

- `README.md`: 项目总体说明与使用方法。
- `requirements.txt`: Python 依赖清单。
- `data/`: 存放原始与处理后数据（包括 qlib 预打包 bin 数据和股票池）。
- `examples/`: 使用示例脚本（例如 `quickstart.py`）。
- `notebooks/`: Jupyter 笔记本集合（当前文件位于此处）。
- `scripts/`: 工具脚本，用于下载与准备数据等。
- `src/`: 项目源码，主要包为 `qlib_project`，包含数据处理与模型运行工具。
